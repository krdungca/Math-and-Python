```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
pnts = 500

theta = np.linspace(0,18*np.pi,pnts)

r1 = np.random.rand()
r2 = np.random.rand()

x = np.cos(r1*theta)
y = np.sin(r2*theta)

plt.plot(x,y,'k-')
plt.plot([x[0],x[-1]],[y[0],y[-1]],'r-')
plt.axis('off')
plt.show()
# plt.plot(theta, x, theta, y)
```
