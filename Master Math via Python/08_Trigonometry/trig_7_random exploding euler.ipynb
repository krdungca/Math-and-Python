```python
import numpy as np

import matplotlib.pyplot as plt
```

Euler's formula/ Euler's identity

"leads to the most elegant equation in all of human civilization"

k = angle

$e^{ik} = cos(k) + i \times sin(k)$

$me^{ik} = m(cos(k) + i \times sin(k))$




```python
k = np.pi / 6
m = 2.3

eul = m * np.exp(1j*k)
cis = m*(np.cos(k) + 1j*np.sin(k))  #cos imag sin = cis


mag = abs(eul) #distance from origin
ang = np.angle(eul)

print(mag)
print(ang)

plt.polar([0,ang],[0,mag],'r')
plt.polar(k,m,'bo')

#plotting custom thetagrids
# plt.thetagrids((350,149,199),labels=None)
plt.thetagrids(range(0,360,15))
plt.show()
```

```python
nvects = 200

ms = np.random.rand(nvects)
ks = np.random.rand(nvects)*2*np.pi      

for i in range(0,nvects):

    r = np.random.rand()

    if r<.4:
        clr = [1,.2,.7]

    elif r>0.4 and r < 0.8:
        clr = [.7,.2,1]

    else:
        clr = [0,1,0]

    plt.polar([0,ks[i]],[0,ms[i]],color=clr)

plt.axis('off')
plt.show()
```
